# Verification of Lemma 3.4 via Gröbner basis elimination (SageMath)

This notebook verifies the elimination step used in Lemma 3.4.


## 1. Define the polynomial ring and the monomial order


```python
# Lexicographic order: r > s > c > d > a > b
R.<r, s, c, d, a, b> = PolynomialRing(QQ, order='lex')
print('Polynomial ring defined successfully:')
print(R)
```

## 2. Define $L_i$ and $R_i$


```python
L1 = r*a - a + s + d
R1 = r + c - a

L2 = s*b + s + d - a
R2 = s + d - a

L3 = a*(b^2 + b + s*b + s + d - a) - b*(a^2 + r*a + s + d)
R3 = a*(b + s + d - a) - b*(r + c)

print('Defined L_i and R_i:')
for name, poly in [('L1', L1), ('R1', R1), ('L2', L2), ('R2', R2), ('L3', L3), ('R3', R3)]:
    print(f'{name} = {poly}')
```

## 3. Define $E_i = (b-a)(L_i-R_i) - R_i(s+d-r-c)$


```python
E1 = (b - a)*(L1 - R1) - R1*(s + d - r - c)
E2 = (b - a)*(L2 - R2) - R2*(s + d - r - c)
E3 = (b - a)*(L3 - R3) - R3*(s + d - r - c)

print('=' * 70)
print('Defined E1, E2, E3:')
print('=' * 70)
print(f'E1 = {E1}')
print(f'E2 = {E2}')
print(f'E3 = {E3}')
```

## 4. Compute the Gröbner basis


```python
print('=' * 70)
print('Computing the Gröbner basis (lex order: r > s > c > d > a > b)...')
print('=' * 70)

I = R.ideal([E1, E2, E3])
G = I.groebner_basis()

print(f'Gröbner basis contains {len(G)} polynomials.')
for i, g in enumerate(G, 1):
    print(f'\nBasis element {i}: {g}')
```

## 5. Extract the elimination polynomial in $\mathbb{Q}[a,b]$


```python
basis_ab = []
for g in G:
    vars_in_g = g.variables()
    if all(v in [a, b] for v in vars_in_g):
        basis_ab.append(g)

print('=' * 70)
print('Polynomials involving only a and b:')
print('=' * 70)

if basis_ab:
    for i, poly in enumerate(basis_ab, 1):
        print(f'\nPolynomial {i}: {poly}')
        print('Factorization:')
        print(factor(poly))
else:
    print('No polynomial involving only a and b was found.')
```

## 6. Verify the expected elimination polynomial


```python
expected = a * b * (a - b)^2

if basis_ab:
    poly = basis_ab[0]
    print('=' * 70)
    print('Verification of the expected elimination polynomial')
    print('=' * 70)
    print(f'Computed polynomial: {poly}')
    print(f'Expected polynomial: {expected}')
    print(f'Match: {poly == expected}')
else:
    print('Run the previous cells first.')
```

## 7. Summary


```python
print('=' * 70)
print('Verification complete')
print('=' * 70)
print('The elimination ideal contains the polynomial P(a,b) = a*b*(a-b)^2.')
print('Under the assumptions a, b > 0 and a != b, this is impossible.')
```
